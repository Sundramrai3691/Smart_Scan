```python
# Import necessary libraries
import requests
import zipfile
from bs4 import BeautifulSoup
import time

# Step 1: Define the Amazon grocery product search URL
search_url = 'https://www.amazon.in/s?k=skincare+product&crid=38TT3G5YP3TP2&sprefix=skincare+product%2Caps%2C259&ref=nb_sb_noss_2'

# Step 2: Set up headers to mimic a browser visit
headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/100.0.4896.75 Safari/537.36"
}

# Function to scrape product images
def scrape_images():
    # Step 3: Send a GET request to the search URL
    try:
        response = requests.get(search_url, headers=headers)

        # Check if the request was successful
        if response.status_code == 200:
            # Step 4: Parse the HTML content with BeautifulSoup
            soup = BeautifulSoup(response.content, 'html.parser')

            # Step 5: Extract product names and image URLs
            product_images = []

            # Find all product containers
            for item in soup.find_all('div', {'data-component-type': 's-search-result'}):
                # Get product image
                img = item.find('img', class_='s-image')
                if img and 'src' in img.attrs:
                    product_images.append(img['src'])

            # Check if any images were found
            if not product_images:
                print("No images found. Please check the scraper setup.")
            else:
                # Step 6: Create a zip file to save images
                zip_file_name = 'amazon_skincare-product_images.zip'
                with zipfile.ZipFile(zip_file_name, 'w') as zipf:
                    # Step 7: Download and save images directly into the zip file
                    for idx, product_image_url in enumerate(product_images):
                        try:
                            # Create a valid filename using a unique index
                            image_filename = f"detergent_image_{idx}.jpg"

                            # Download the image
                            img_data = requests.get(product_image_url).content

                            # Check if the response status is OK
                            if img_data:
                                # Add the image to the zip file
                                zipf.writestr(image_filename, img_data)
                                print(f"Downloaded and added to zip: {image_filename}")
                            else:
                                print(f"Failed to download image from {product_image_url}")

                        except Exception as e:
                            print(f"Failed to download image: {str(e)}")

                print(f"Images have been zipped into: {zip_file_name}")

        elif response.status_code == 503:
            print("Server is temporarily unavailable (503). Retrying...")
            time.sleep(5)  # Wait for a bit before retrying
            scrape_images()  # Retry the scraping

        else:
            print(f"Failed to retrieve products. Status code: {response.status_code}")

    except Exception as e:
        print(f"An error occurred: {str(e)}")

# Run the scrape_images function
scrape_images()
```
